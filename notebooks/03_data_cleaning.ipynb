### 🧹 Data Cleaning: Step 1

Investigate the 13 Exact Duplicate Rows

🎯 Objective

Before removing duplicates, we need to determine:

- How many duplicate rows exist?
- Are the duplicated records completely identical?
- Do they have the same target label?
- Could removing them affect the class distribution?

Important: We won't modify the original dataset. All investigations will happen on a working DataFrame.

```python
# Import libraries
from pathlib import Path
import pandas as pd
import numpy as np

# Load the original dataset
DATA_PATH = Path("../data/raw/meajor_dataset.csv")
 
df = pd.read_csv(DATA_PATH)

# Display dataset dimensions
print("Dataset shape:", df.shape)

# Display column names
print("\nColumns:")
print(df.columns.tolist())

# Display the first five rows
display(df.head())
```

```python
# Verify the duplicated rows
duplicate_mask = df.duplicated(keep=False)

print("Total rows:", len(df))
print("Rows involved in exact duplicates:", duplicate_mask.sum())
print("Duplicate rows after the first occurrence:", df.duplicated().sum())
```

```python
# Inspect the duplicated rows
# Display all rows involved in exact duplicates
duplicate_rows = df[duplicate_mask].sort_values(
    by=df.columns.tolist()
)

display(duplicate_rows)
```

```python
# Count the number of occurrences of each exact row
duplicate_groups = (
    df[df.duplicated(keep=False)]
    .groupby(df.columns.tolist(), dropna=False)
    .size()
    .reset_index(name="occurrences")
    .sort_values("occurrences", ascending=False)
)

print("Number of duplicate groups:", len(duplicate_groups))
print("\nOccurrences per duplicate group:")
display(duplicate_groups[["occurrences", "label"]]
        .value_counts()
        .rename("number_of_groups")
        .reset_index())
```

```python
# Examine the labels of rows involved in duplicate groups
duplicate_label_counts = (
    df.loc[duplicate_mask, "label"]
    .value_counts(dropna=False)
)

display(duplicate_label_counts)
```

```python
# Remove the exact duplicate rows, keeping the original dataframe
# Preserve the original dataset in memory
df_original = df.copy()

# Create a separate working DataFrame
df_cleaned = df.copy()

# Remove exact duplicate rows, retaining the first occurrence
df_cleaned = df_cleaned.drop_duplicates().reset_index(drop=True)

print("Original shape:", df_original.shape)
print("Shape after duplicate removal:", df_cleaned.shape)
print("Rows removed:", len(df_original) - len(df_cleaned))
```

### 🧹 Step 2: Handle the Missing Target Label

🎯 Objective

Your EDA identified one record with a missing target label. Since our goal is to build a spam and phishing detector, we need to investigate that record before deciding what to do.

```python
# Count the missing target labels
# Check missing values in the target column
print("Missing target labels:", df_cleaned["label"].isna().sum())
```

```python
# Display records with missing target labels
missing_label_rows = df_cleaned[df_cleaned["label"].isna()]

display(missing_label_rows.T)
```

```python
# Remove the record if its label cannot be recovered
# Remove records with missing target labels
before = len(df_cleaned)

df_cleaned = (
    df_cleaned
    .dropna(subset=["label"])
    .reset_index(drop=True)
)

removed = before - len(df_cleaned)

print("Rows removed:", removed)
print("Updated dataset shape:", df_cleaned.shape)
print("Missing target labels remaining:", df_cleaned["label"].isna().sum())
```

## 🧹 Step 3: Investigate Suspicious Timestamps

Your EDA identified timestamps ranging from 1889 to 2102. Before changing them, we need to determine whether they're invalid, incorrectly parsed, or simply unusual records.

Our rule: Investigate first, modify second. We don't want to delete useful data just because a date looks strange.

```python
# Identify the timestamp columns
# Identify columns containing timestamp or date information
date_columns = [
    col for col in df_cleaned.columns
    if any(keyword in col.lower() for keyword in ["date", "time", "timestamp"])
]

print("Potential timestamp columns:")
print(date_columns)

df_cleaned[date_columns].info()
```

```python
# Inspect the suspicious timestamp 
# Replace 'timestamp' with your actual timestamp column name
timestamp_col = "date"

print("Data type:", df_cleaned[timestamp_col].dtype)

display(
    df_cleaned[
        df_cleaned[timestamp_col].astype(str).str.contains(
            r"\b(18\d{2}|19\d{2}|20\d{2}|21\d{2})\b",
            regex=True,
            na=False
        )
    ][[timestamp_col]].head(10)
)

# Check the timestamp range
# Inspect the earliest and latest timestamps
print("Earliest timestamp:", df_cleaned[timestamp_col].min())
print("Latest timestamp:", df_cleaned[timestamp_col].max())
```

```python
# Identify the suspicious timestamps correctly
# Parse dates for investigation without modifying df_cleaned
parsed_dates = pd.to_datetime(
    df_cleaned["date"],
    errors="coerce",
    utc=True
)

# Identify dates outside our provisional investigation range
suspicious_mask = (
    parsed_dates.notna()
    & (
        (parsed_dates.dt.year < 1900)
        | (parsed_dates.dt.year > 2026)
    )
)

print("Earliest timestamp:", parsed_dates.min())
print("Latest timestamp:", parsed_dates.max())

print(
    "Unparseable timestamps:",
    parsed_dates.isna().sum()
)

print(
    "Records before 1900 or after 2026:",
    suspicious_mask.sum()
)
```

```python
# Inspect the actual suspicious records
# Select suspicious records for investigation
suspicious_records = df_cleaned.loc[
    suspicious_mask,
    ["date", "source", "sender_domain", "subject", "label"]
].copy()

display(
    suspicious_records.sort_values("date")
)

print("Suspicious records:", len(suspicious_records))
```

```python
# Check whether the anomalies are concentrated in particular sources
# Create a temporary investigation DataFrame
date_analysis = df_cleaned.copy()

date_analysis["_parsed_date"] = pd.to_datetime(
    date_analysis["date"],
    errors="coerce",
    utc=True
)

# Flag timestamps outside the provisional range
date_analysis["_suspicious_date"] = (
    (date_analysis["_parsed_date"].dt.year < 1900)
    | (date_analysis["_parsed_date"].dt.year > 2026)
)

# Compare suspicious and non-suspicious records by source
source_date_summary = (
    date_analysis.groupby("source")
    .agg(
        total_records=("_suspicious_date", "size"),
        suspicious_records=("_suspicious_date", "sum")
    )
)

source_date_summary["suspicious_percentage"] = (
    source_date_summary["suspicious_records"]
    / source_date_summary["total_records"]
    * 100
)

display(
    source_date_summary.sort_values(
        "suspicious_records",
        ascending=False
    )
)
```

```python
# Inspect the suspicious dates by year
# Extract the year from the parsed timestamps
date_analysis["_year"] = (
    date_analysis["_parsed_date"].dt.year
)

# Summarize the suspicious years
suspicious_year_counts = (
    date_analysis.loc[
        date_analysis["_suspicious_date"],
        "_year"
    ]
    .value_counts()
    .sort_index()
)

display(suspicious_year_counts)
```

```python
# Columns to inspect
context_columns = [
    "date",
    "source",
    "sender_domain",
    "subject",
    "label"
]

# Select suspicious records, include the parsed date for sorting,
# then display only the desired context columns
suspicious_records = date_analysis.loc[
    date_analysis["_suspicious_date"],
    context_columns + ["_parsed_date"]
].sort_values("_parsed_date")

display(
    suspicious_records[context_columns].head(20)
)
```

```python
# Check the suspicious timestamps by source

# Summarize suspicious timestamps by source
source_date_summary = (
    date_analysis.groupby("source")
    .agg(
        total_records=("_suspicious_date", "size"),
        suspicious_records=("_suspicious_date", "sum")
    )
)

source_date_summary["suspicious_percentage"] = (
    source_date_summary["suspicious_records"]
    / source_date_summary["total_records"]
    * 100
)

display(
    source_date_summary.sort_values(
        "suspicious_records",
        ascending=False
    )
)
```

```python
# Compare target-label distributions
label_date_summary = (
    date_analysis.groupby("_suspicious_date")["label"]
    .value_counts(normalize=True)
    .unstack(fill_value=0)
    .mul(100)
)

label_date_summary.index = [
    "Normal timestamps",
    "Suspicious timestamps"
]

display(label_date_summary)
```

```python
# Count benign and phishing records by source and timestamp status
source_label_summary = (
    date_analysis
    .groupby(["source", "_suspicious_date", "label"])
    .size()
    .unstack("label", fill_value=0)
    .rename(columns={
        0.0: "benign",
        1.0: "phishing"
    })
)

display(source_label_summary)
```

```python
# Inspect suspicious timestamps by year and source
suspicious_date_details = date_analysis.loc[
    date_analysis["_suspicious_date"],
    ["date", "source", "_parsed_date", "_year", "label"]
].copy()

display(
    suspicious_date_details
    .groupby(["source", "_year"])
    .size()
    .unstack(fill_value=0)
)
```

### Timestamp Anomaly Investigation

The dataset contains 218 timestamps outside the provisional investigation range of 1900–2026. These records are distributed across the TREC-05, TREC-06, and TREC-07 sources, and all have the phishing label.

A comparison with the original dataset confirmed that the examined records contain the same timestamps. Therefore, the anomalies were not introduced by the current preprocessing steps for those records.

However, the historical accuracy of the timestamps could not be independently verified. No timestamps or records were modified or removed.

**Decision:** Preserve the original timestamps and labels. Exclude the date feature from the initial model inputs pending further investigation into its reliability and predictive value.


```python
# Preserve the cleaned dataset
df_cleaned = df.copy()

# Verify that the date and label columns are present
print("Date column exists:", "date" in df_cleaned.columns)
print("Label column exists:", "label" in df_cleaned.columns)
```

```python
# Create a separate DataFrame for initial model development
df_model = df_cleaned.drop(columns=["date"]).copy()

# Verify the changes
print("Cleaned dataset shape:", df_cleaned.shape)
print("Model dataset shape:", df_model.shape)

print("\nDate retained in cleaned dataset:", "date" in df_cleaned.columns)
print("Date included in model dataset:", "date" in df_model.columns)
print("Label retained in model dataset:", "label" in df_model.columns)
```

```python
# Create the processed data directory
PROCESSED_DIR = Path("../data/processed")
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Save the cleaned dataset
df_cleaned.to_csv(
    PROCESSED_DIR / "meajor_cleaned.csv",
    index=False
)

# Save the model dataset without the date feature
df_model.to_csv(
    PROCESSED_DIR / "meajor_model.csv",
    index=False
)

print("Datasets saved successfully!")

print("Cleaned dataset:", df_cleaned.shape)
print("Model dataset:", df_model.shape)
```
